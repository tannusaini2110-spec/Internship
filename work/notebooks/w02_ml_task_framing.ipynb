## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
print("Task type: Classification (binary).")
print()
print("I am predicting whether a page falls into the declining category or not")
print("-- a yes/no label per page, not a numeric score or an undefined group.")
print("This is not clustering, since I already know the categories I care about:")
print("declining vs stable/growing. It is not pure ranking either, since I need")
print("a clear flag rather than a full ordered list.")
print()
print("In practice this could also be framed as scoring (a decline-risk probability),")
print("but the underlying task is a classification problem.")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
print("""
Target: is_declining -- a binary label built from trend_direction.
is_declining = 1 if trend_direction == 'down', else 0.

This is a PROXY, not a perfectly clean ground truth: 'declining' here means
the recent trend in impressions/clicks is downward, based on the data
already computed in the dataset. It's an observed outcome (calculated from
real traffic numbers), not a hand-defined rule -- but it's still a proxy
for the deeper question of 'is this page truly losing relevance,' since
short-term dips can have many causes (seasonality, algorithm noise, etc).
""")
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
print("Success metric: Precision@K (e.g. Precision@50).")
print()
print("This measures: of the top K pages my model flags as most likely to be")
print("declining, what fraction are actually declining? This fits the real")
print("action -- the content team only has bandwidth to review a limited")
print("shortlist each week, so what matters is whether the TOP flagged pages")
print("are correct, not overall accuracy across all pages.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
print("A fixed rule (e.g. 'flag anything stale AND low-traffic') is too rigid:")
print("- It can't weigh multiple weak signals together (position, freshness,")
print("  CTR, word count) the way a model can.")
print("- The real relationship between these signals and decline isn't a clean")
print("  cutoff -- our own earlier test showed freshness alone doesn't predict")
print("  CTR in a simple linear way (91-180 days actually scored HIGHER than")
print("  fresher content).")
print("- A model can learn these messy, non-linear interactions and rank pages")
print("  by risk, instead of a rigid yes/no threshold that misses edge cases.")
```
